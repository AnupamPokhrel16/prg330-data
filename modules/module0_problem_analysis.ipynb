### Module 0: Problem Analysis

Before writing any code, we defined exactly what we were trying to find out and got familiar with
the dataset we would be using. This module covers our objective, the dataset itself, and the
problem statement the rest of the notebook is built to answer.


#### Objective

**In one line:** find out which city has the worst air quality, why, and what's driving it -- using
a full year of real hourly data across six cities.

- Rank the six cities by air quality and measure how big the gap actually is.
- Check whether pollution follows a predictable pattern across the year.
- Find out whether the dataset's worst hours are a trend or a one-off event.
- Identify which pollutants most strongly drive the overall AQI.
- Decide, honestly, what to do with a pollutant column (CO2) that's missing for most of the year.

The detailed version of each question, and why it matters, is in the Problem Statement below.


#### The dataset

`Air_Quality.csv` contains 52,704 hourly air-quality readings -- one row per city per hour, for all
of 2024 (366 days &times; 24 hours &times; 6 cities = 52,704 rows exactly, since 2024 is a leap
year).

**Cities covered:** Brasilia, Cairo, Dubai, London, New York, Sydney -- six cities spread across
six continents, chosen to compare very different climates and pollution sources rather than a
single region.

**Columns:**

| Column | What it is | Unit |
| :--- | :--- | :--- |
| `Date` | Timestamp of the reading | hourly timestamp |
| `City` | Which of the six cities the row belongs to | -- |
| `CO` | Carbon Monoxide level | ppb |
| `CO2` | Carbon Dioxide level (only recorded from 26 Oct 2024 onward -- see Module 6) | ppm |
| `NO2` | Nitrogen Dioxide level | ppb |
| `SO2` | Sulfur Dioxide level | ppb |
| `O3` | Ozone level | ppb |
| `PM2.5` | Fine particulate matter | &micro;g/m&sup3; |
| `PM10` | Coarse particulate matter | &micro;g/m&sup3; |
| `AQI` | Air Quality Index -- the EPA-style overall score the rest of this project is built around | 0-500 index |

AQI is the column every question in the Problem Statement below ultimately comes back to: it is the
single number that summarizes all seven pollutant readings into one comparable score per hour,
which is why the city, month, and pollutant comparisons later in this notebook all use it as the
common yardstick.


#### Problem statement

Air pollution is one of the largest environmental health risks worldwide, and the level of risk
varies sharply by city, season, and pollutant. This project analyzes one year (2024) of hourly
air-quality data from six cities -- Brasilia, Cairo, Dubai, London, New York, and Sydney -- to
answer four concrete questions:

1. **Which cities have the worst and best air quality, and by how much?**
2. **Does air quality change predictably across the year, and if so, when is it worst?**
3. **Are the dataset's worst pollution hours spread evenly across the year, or concentrated in
   specific events?**
4. **Which pollutants are the strongest drivers of the overall Air Quality Index (AQI), and how
   strong is that relationship?**

We also had to resolve a data-quality problem discovered while exploring the dataset: the `CO2`
column is missing for 81.7% of all rows. Before using it in any conclusion, we needed to determine
*why* it was missing -- a sensor outage, inconsistent reporting, or a feature that simply was not
being recorded yet -- because each of those has a different, correct way to handle it.

**Why this matters.** AQI drives real decisions -- outdoor activity advisories, school closures,
industrial regulation -- so getting the city ranking, seasonal pattern, and pollutant drivers right
(and being honest about what the data can and cannot support) has consequences beyond this
notebook.
